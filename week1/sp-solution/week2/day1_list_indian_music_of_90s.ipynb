### Using Groq LLM Model 

```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
# Connect to OpenAI client library
# A thin wrapper around calls to HTTP endpoints

openai = OpenAI()

# For Gemini, DeepSeek and Groq, we can use the OpenAI python client
# Because Google and DeepSeek have endpoints compatible with OpenAI
# And OpenAI allows you to change the base_url

anthropic_url = "https://api.anthropic.com/v1/"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
deepseek_url = "https://api.deepseek.com"
groq_url = "https://api.groq.com/openai/v1"
grok_url = "https://api.x.ai/v1"
openrouter_url = "https://openrouter.ai/api/v1"
ollama_url = "http://localhost:11434/v1"




```

```python
# Check if Groq API ke present or not
load_dotenv(override=True)
sp_ai_api_key_groq=os.getenv('SP_AI_API_KEY_GROQ')

if sp_ai_api_key_groq:
    print("Groq API key present")
else:
    print("Error: No api key for GroK")   
```

```python
#Groq API call

groq =OpenAI(api_key=sp_ai_api_key_groq,base_url=groq_url)
ollama=OpenAI(api_key="ollama",base_url=ollama_url)
```

```python
ques="""Tell me the list of Indian song you like to listen of 90's  """
puzzle=[{"role":"user","content":ques},]
tel_me_a_joke=[{"role":"user","content":"Please tell me your name"}]
```

```python
response = groq.chat.completions.create(model="openai/gpt-oss-120b",messages=puzzle)
display(Markdown(response.choices[0].message.content))
```

# Calling the Gemini LLM model


```python
#Load API Key
load_dotenv(override=True)
gemini_api_key = os.getenv('sp_ai_api_Key_Gemini')

if gemini_api_key:
    print("Gemmi API Key present")
else:
    print("Error: No Gemini API key found")    
```

```python
# call Gemini API model

gemini= OpenAI(api_key=gemini_api_key,base_url=gemini_url)


```

```python
# Response generated from Gemini LLM models

response=gemini.chat.completions.create(model="gemini-2.5-pro",messages=puzzle)
display(Markdown(response.choices[0].message.content))
```
